# Health apps dataset (Google Fit & Apple HealthKit)

### Description

Smartphone apps and wearable devices for monitoring physical activity and other health behaviors have gained popularity in recent years, offering a largely untapped source of data on health behaviors in real-world settings. This data is vast, collected at a low cost in natural environments, and often recorded automatically, making it a powerful supplement to traditional surveillance studies and controlled trials. These datasets are uncovering new insights into the environmental and social factors influencing physical activity. However, the observational nature of the data and its collection through commercial devices and apps present challenges, such as the potential for measurement, population, or selection bias, as well as issues with missing data.

This dataset includes anonymized health and fitness data gathered from users of Google Fit and Apple HealthKit. It encompasses various categories such as body measurements (weight, height), physical activity (steps, exercise duration), nutrition, sleep patterns, heart rate, and cardiovascular health metrics. Apple HealthKit extends data collection to include more extensive medical and health-related information. Both platforms collect data through wearable devices and self-reported inputs, providing a comprehensive view of users' health and lifestyles.

### Introduction

**Google Fit** is a fitness tracking platform that was released in October 2014. It is an open ecosystem that currently supports Android 4.1 and higher. It reads, gathers, and stores users' data obtained via wearable gadgets and self-reporting.

**HealthKit** (Apple Health) is Apple's health data collection and visualization platform that was released in 2014 for iPhone with iOS 8+. The Health app collects and visualizes user data received from Apple Watch, smartphones, and self-reported inputs. HealthKit is a framework with an API that allows displaying the health data collected by apps and wearables that are connected to the Health app.

#### Key Differences

- **Platform compatibility:** Google Fit works with any platform, whereas Health is an iOS-only app (therefore, data from Health, specifically, is for iOS devices only).

- **Clinical use:** Google Fit APIs are meant for creating fitness apps and should not be used for creating clinical solutions. Google Fit cannot be used with medical devices or as a digital therapeutic. HealthKit is built to help develop digital therapeutics — apps that can be used for patients' treatment.

- **Data storage:** Google Fit stores data in the cloud, whereas Health and HealthKit save it on the user's device.

#### Data Collection Methods

- **Google Fit:** Gathers data through wearable devices and self-reporting, supporting a variety of devices and platforms, and stores user data in the cloud for synchronization across devices.

- **Apple HealthKit:** Collects data from Apple Watch sensors and self-reported inputs, exclusive to iOS devices, and saves data locally on the user's device to enhance privacy and security.

#### Data Categories

We have split the data into the following categories:

- **Cardiovascular health:** Heart rate, heart rate variability, blood pressure, VO2 max
- **Vital signs:** Body temperature, oxygen saturation, respiratory rate
- **Physical activity:** Steps, distance, active energy burned, exercise duration, workouts
- **Body metabolism:** Basal metabolic rate, energy expenditure
- **Sleep & Mindfulness:** Sleep duration, sleep stages, mindful minutes
- **Nutrition:** Dietary intake, water consumption, macronutrients
- **Anthropometrics:** Weight, height, body mass index (BMI), body fat percentage

### Data Quality and Sanitation

Quality control for wearable data focuses on ensuring complete and reliable health metrics with minimal gaps or anomalies. Automated checks are performed to identify:

- Missing data periods or gaps in continuous monitoring
- Out-of-range values (e.g., physiologically impossible heart rates or step counts)
- Inconsistencies between data sources
- Device connectivity issues or sync failures

QC status is stored in metadata, and problematic data periods are flagged for review or exclusion from downstream analysis.

#### Scheduled Data Quality Checks Recommendations

For each new batch of data acquired and added to the database:

- All recommended checks in the "Data Sanitation and Quality Controls"
- Temporal distribution shift for tabular features
- Data completeness report
- Cross-validation between Google Fit and Apple Health data sources

### Known Issues and Limitations

- **Measurement bias:** Potential for measurement, population, or selection bias due to the observational nature of the data and collection through commercial devices
- **Missing data:** Gaps in data collection due to device connectivity issues, battery depletion, or user non-compliance with wearing devices
- **Device variability:** Different wearable devices may have varying levels of accuracy for the same metrics
- **Self-reported data:** Nutrition and some other metrics rely on user input, which may be incomplete or inaccurate
- **Platform differences:** Google Fit and Apple Health have different data structures and feature sets, requiring careful harmonization

### Association with Clinical Factors

Current research in wearable health data has shown associations between device-collected metrics and various health outcomes:

- **Physical activity:** Steps and active minutes correlate with cardiovascular health, metabolic health, and overall mortality
- **Sleep patterns:** Sleep duration and quality are associated with metabolic disorders, cardiovascular disease, and mental health
- **Heart rate variability:** HRV is linked to stress levels, autonomic nervous system function, and cardiovascular health
- **Resting heart rate:** Elevated resting heart rate is associated with increased cardiovascular risk
- **Body composition:** Weight and BMI trends correlate with metabolic health outcomes
